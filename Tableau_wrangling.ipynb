## 2. Data Preprocessing

### Insights from Dataset Exploration

Several preprocessing steps were identified after reviewing the dataset structure, missing values, and data quality:

- The POI dataset contained one row with missing values, which was removed to maintain data consistency.
  
- The Amenities dataset contained multiple missing (NaN) values. As these values likely indicate the absence of an amenity, they were replaced with 0 to represent that the amenity does not exist for the property.
  
- The PropertyNumberOfFloors column in the Sales Transactions dataset contained more than 50% missing values. Due to the high proportion of incomplete data, the column was removed from the analysis.
  
- The NumberOfBedroomsInTransactedUnit column also contained a significant number of missing values. Missing entries were replaced with -1 to indicate unknown bedroom counts while preserving the records for analysis.

- Missing values in PropertyYearOfCompletion and PropertynumberOfUnits were filled using the median values of their respective columns to reduce the impact of missing data while maintaining the overall distribution of the dataset.

```python
# Drop duplicates
POI = POI.drop_duplicates()
Amenities = Amenities.drop_duplicates()
SM_assets = SM_assets.drop_duplicates()
Sales = Sales.drop_duplicates()
Subzones = Subzones.drop_duplicates()

# Drop row from POI
POI = POI.dropna(subset=['POI_Latitude', 'POI_Longitude'])

# Fill NaN with 2 (does not exist)
Amenities = Amenities.fillna(2)

# Drop high-missing column
Sales = Sales.drop(columns=['PropertyNumberOfFloors'])

# Fill bedroom missing values with -1 (Unknown)
Sales['NumberOfBedroomsInTransactedUnit'] = (
    Sales['NumberOfBedroomsInTransactedUnit'].fillna(-1)
)

# Fill columns with median
Sales['PropertyYearOfCompletion'] = (
    Sales['PropertyYearOfCompletion'].fillna(
        Sales['PropertyYearOfCompletion'].median()
    )
)

Sales['PropertynumberOfUnits'] = (
    Sales['PropertynumberOfUnits'].fillna(
        Sales['PropertynumberOfUnits'].median()
    )
)
```

## 3. Dataset Merging

The datasets were merged into a single consolidated dataset to support further analysis and Tableau visualisation. This will make it easier to create visualisations in Tableau without having to manually merge the datasets in the software itself.

- To merge the datasets, the key column PropertyID will be used to first merge the Sales, Amenities and SM_assets dataset.

- To merge the final POI and Subzone datasets, SubzoneId is used. This will create the final dataset to be used in Tableau. In addition, a new column will be created to count the number of POIs in the vicinity (by subzones) of each property.

- To prepare for dataset merging, final checks and data standardization will be carried out first.

### All Datasets are filled with no missing values, especially for PropertyID, which will be used as the key to merge Sales, Amenities and SM_assets first.

```python
# Merge first three datasets
Sales_Amenities = Sales.merge(Amenities, on='PropertyID', how='left')
Sales_Full = Sales_Amenities.merge(SM_assets, on='PropertyID', how='left')
```

```python
# Check new dataset
Sales_Full.isnull().sum()
```

there are multiple NaN values for some columns which will be resolved later.

#### For the last two datasets, Subzones will be merged first as POI relies on the Subzone due to the hierarchical structure of the data. The Property dataset links each property to a specific subzone. To match POIs to the relevant properties, The POI dataset uses the subzone ID to identify which properties should link to which POIs.

```python
# Merge Subzones first 
Sales_Full = Sales_Full.merge(
    Subzones,
    left_on='PropertyInSubzone',
    right_on='SubzoneId',
    how='left'
)
```

```python
# create new column for POI count (used for Tableau visulisation)
POI_features = POI.pivot_table(
    index='POI_InSubzone',
    columns='POI_Type',
    aggfunc='size',
    fill_value=0
).reset_index()

# Merge POI last
Sales_Full = Sales_Full.merge(
    POI_features,
    left_on='PropertyInSubzone',
    right_on='POI_InSubzone',
    how='left'
)
```

```python
# Check Master Dataset
Sales_Full.info()
```

#### With all the datasets merged, missing values can now be dealt with.

```python
# Reformatting of certain amenities columns

# Check unique values for object columns
cols = ['Golf','Cafe','Salon','Sauna','Shop']

for col in cols:
    print(col, Sales_Full[col].unique())


for col in cols:
    
    nan_count = (Sales_Full[col] == 'nan').sum()
    
    nil_count = (Sales_Full[col] == 'NIL').sum()
    
    print(f"{col}")
    print(f"NaN Count: {nan_count}")
    print(f"NIL Count: {nil_count}")
    print()
```

```python
# Replace with integers
cols = ['Golf','Cafe','Salon','Sauna','Shop']

Sales_Full[cols] = Sales_Full[cols].replace('NIL', 2)
Sales_Full[cols] = Sales_Full[cols].replace('nan', 2)
```

```python
# Reformat into float64
cols = ['Golf','Cafe','Salon','Sauna','Shop']
Sales_Full[cols] = Sales_Full[cols].astype(float)
```

```python
# Check reformatting success
Sales_Full.info()
```

```python
# Fill NaN with 2 (data unavailable)
cols = ['Gym','Pool','Playground','Tennis','Golf','Cafe','LapPool','KidsPool','Squash','Badminton','Basketball','BBQ','Jacuzzi','Security','Salon','Sauna','Shop']
Sales_Full[cols] = Sales_Full[cols].fillna(2)
```

```python
# Fill NaN with median of column
sm_cols = [
    'NumberofSchoolsWithin500Mts',
    'NumberOfMRTsWithin500Mts',
    'NumberOfBusStopsWithin500Mts',
    'TimeToRafflesByCarInSeconds',
    'TimeToRafflesByPublicTransportInSeconds',
    'TimeToAirportByCarInSeconds',
    'TimeToAirportByPublicTransportInSeconds'
]

for col in sm_cols:
    Sales_Full[col] = Sales_Full[col].fillna(Sales_Full[col].median())
```
